```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
results = {
    "Optimizer": [
        "GD with Momentum",
        "SGD with Momentum",
        "Mini-Batch GD with Momentum",
        "RMSProp with Momentum",
        "Adam"
    ],
    
    "Training Loss": [
        0.0612,
        0.0352,
        0.0569,
        0.0384,
        0.0941
    ],
    
    "Test Loss": [
        0.0567,
        0.0930,
        0.0553,
        0.0745,
        0.0807
    ],
    
    "Accuracy": [
        0.9825,
        0.9649,
        0.9912,
        0.9825,
        0.9912
    ],
    
    "Precision": [
        0.9859,
        0.9855,
        0.9861,
        0.9859,
        0.9861
    ],
    
    "Recall": [
        0.9859,
        0.9577,
        1.0000,
        0.9859,
        1.0000
    ],
    
    "F1 Score": [
        0.9859,
        0.9714,
        0.9930,
        0.9859,
        0.9930
    ]
}

results_df = pd.DataFrame(results)

results_df
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    results_df["Optimizer"],
    results_df["Accuracy"]
)

plt.xlabel("Optimizer")
plt.ylabel("Accuracy")
plt.title("Accuracy Comparison")
plt.xticks(rotation=20)
plt.ylim(0.9, 1.0)
plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    results_df["Optimizer"],
    results_df["Precision"]
)

plt.xlabel("Optimizer")
plt.ylabel("Precision")
plt.title("Precision Comparison")
plt.xticks(rotation=20)
plt.ylim(0.9, 1.0)
plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    results_df["Optimizer"],
    results_df["Recall"]
)

plt.xlabel("Optimizer")
plt.ylabel("Recall")
plt.title("Recall Comparison")
plt.xticks(rotation=20)
plt.ylim(0.9, 1.0)
plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    results_df["Optimizer"],
    results_df["F1 Score"]
)

plt.xlabel("Optimizer")
plt.ylabel("F1 Score")
plt.title("F1 Score Comparison")
plt.xticks(rotation=20)
plt.ylim(0.9, 1.0)
plt.grid(axis="y")

plt.show()
```

```python
x = np.arange(len(results_df))
width = 0.35

plt.figure(figsize=(10, 5))

plt.bar(
    x - width/2,
    results_df["Training Loss"],
    width,
    label="Training Loss"
)

plt.bar(
    x + width/2,
    results_df["Test Loss"],
    width,
    label="Test Loss"
)

plt.xlabel("Optimizer")
plt.ylabel("Loss")
plt.title("Training Loss vs Test Loss Comparison")
plt.xticks(
    x,
    results_df["Optimizer"],
    rotation=20
)

plt.legend()
plt.grid(axis="y")

plt.show()
```

```python
metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]

x = np.arange(len(results_df))
width = 0.18

plt.figure(figsize=(12, 6))

for i, metric in enumerate(metrics):
    plt.bar(
        x + (i - 1.5) * width,
        results_df[metric],
        width,
        label=metric
    )

plt.xlabel("Optimizer")
plt.ylabel("Score")
plt.title("Overall Performance Comparison")
plt.xticks(
    x,
    results_df["Optimizer"],
    rotation=20
)

plt.ylim(0.9, 1.0)
plt.legend()
plt.grid(axis="y")

plt.show()
```

```python
best_optimizer = results_df.loc[
    results_df["F1 Score"].idxmax()
]

print("Best Optimizer based on F1 Score")
print("--------------------------------")
print("Optimizer:", best_optimizer["Optimizer"])
print(f"Accuracy:  {best_optimizer['Accuracy']:.4f}")
print(f"Precision: {best_optimizer['Precision']:.4f}")
print(f"Recall:    {best_optimizer['Recall']:.4f}")
print(f"F1 Score:  {best_optimizer['F1 Score']:.4f}")
print(f"Test Loss: {best_optimizer['Test Loss']:.4f}")
```
